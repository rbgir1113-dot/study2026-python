# 값을 몇 개나 전달할 지 모를 때

```python
# 값을 몇 개를 전달해도 모두 더하는 함수
# args : 가변인자, 가변매개변수, rest parameter

def total(*args):
    print(args)
    print(type(args))
    return sum(args)
```

```python
total(10, 20)
```

```python
# 두 수를 전달하면 사칙연산의 결과 반환
def calc(num1, num2):
    return num1 + num2 , num1 - num2 , num1 * num2, num1 / num2
```

```python
# 언패킹
add, sub, mul, div = calc(10,20)
```

```python
print(add)
print(sub)
print(mul)
print(div)
```

```python
# 국어, 영어, 수학 점수 평균 구하기
# 자릿수는 2번째 자리부터 반올림
# 사용자의 정보가 입력되면 같이 출력하기
def get_average(**kwards):
    kor, eng, math = kwards.get("kor", 0), kwards.get("eng", 0), kwards.get("math", 0)
    total = kor + eng + math

    average = total / 3
    
    if "round" in kwards:
        average = round(average, kwards["round"])
    
    if "name" in kwards:
        name = kwards["name"]
        return f"{name}님의 평균은: {average}"

    return f"평균은: {average}"
```

```python
user_info = {
    "kor": 100,
    "eng": 80,
    "math": 22,
    "name": "홍길동",
    "round": 5
}
```

```python
get_average(**user_info)
```
