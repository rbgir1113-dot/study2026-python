# 8. 삼항 연산자

```python
bool("True" if 11 == 10 else "False")
```

```python
# 삼항 두수 비교
# 들여쓰기로 영역을 구분한다.
num1 = input("첫번째 정수")
num2 = input("두번째 정수")
```

```python
num1 if num1 > num2 else "두 수가 같습니다" if num1 == num2 else num2
```

```python
# 사용자에게 정수 또는 실수를 입력 받아서
# 정수 인지 실수 인지 확인후 삼항 연산자로 결과를 출력하시오.
# ex) 10 -> "정수"
# ex) 11.0 -> "정수"
# ex) 11.3 -> "실수"
```

```python
num = input("수를 입력하세요")
```

```python
"정수" if float(num)%1 == 0 else "실수"
```
