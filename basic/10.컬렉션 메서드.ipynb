# 1. 리스트 컬렉션 메서드

```python
li1 = [i +1 for i in range(10)]
```

```python
# 맨 뒤에 추가
li1.append(10)
```

```python
li1
```

```python
li1 = [i for i in range(1, 21)]
```

```python
li1
```

```python
# list의 모든 합 구하기
# sum() : 모든 값을 더하는 함수
# min() : 최솟값
# max() : 최대값

print(sum(li1))
print(min(li1))
print(max(li1))
```
