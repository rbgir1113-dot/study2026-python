# 컴프리헨션

```python
for i in [10,20,30,40]:
    print(i)
```

```python
[i * 2 for i in [10,20,30]]
```

```python
{i * 2 for i in [10,20,30]}
```

```python
name = "장보고", "홍길동", "이순신"
```

```python
for i in name:
    print(i)
```

```python
# 인덱스를 만들던가
# [(0,장보고), (1,홍길동), (2,이순신)]
# unpacking

# enumerate(name)
for i in enumerate(name):
    print(i)
    


```

```python
for index, name in enumerate(name):
    print(f"{index+1}번 {name}")
```

```python
# 일반 for 문
# range(10) -> ~9
# range(10,20) -> 10~19
# range(20,30,2) -> 20~29, 2칸씩

for i in range(10):
    print(i)
```

```python
# 1~50까지 홀수만 출력하기
for i in range(1,51,2):
    print(i)
    
```

```python
# 1 ~ 50까지 홀수만 리스트에 담아서 출력
# 컴프리헨션 문법 사용
print([i for i in range(1, 51, 2)])
```

```python
# 1 ~ 50까지 홀수만 리스트에 담아서 출력
# 세트 컴프리헨션에 0,1,2,3,4,5

{i % 6 for i in range(50)}
```

# 딕셔너리 빠른 for 문

```python
foods = {
    "a":"카레",
    "b":"라면",
    "c":"불고기",
    "d":"떡볶이"
}
```

```python
for i in foods:
    print(foods.get(i))
```

```python
for i in foods.values():
    print(i)
```

```python
for i in list(foods.values()):
    print(i)
```

```python
for key, food in foods.items():
    print(key,food)
```
