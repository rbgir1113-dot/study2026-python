```python
# 완전 객체 지향형 언어
# Market
# 판매
class Market:
    def __init__(self, market_name, product_name, product_price):
        self.market_name = market_name
        self.product_name = product_name
        self.product_price = product_price

    def sell(self, customer):
        customer.money -= self.product_price * ( 100 - customer.discount) / 100


# Customer
# 할인율
class Customer:
    def __init__(self, name, money, discount):
        self.name = name
        self.money = money
        self.discount = discount
```

```python
cu = Market("CU", "도시락",10000)

customer = Customer("이규혁", 50000, 10)

cu.sell(customer)
print(customer.money)
```

```python
# Atm, User 객체
# deposit

# Atm.deposit(User, 10000)
# 유저의 돈은 10000 차감,
# 은행에 돈은 10000 입금

# 은행이 가진돈과 유저의 돈 출력
```

```python
class Atm:
    def __init__(self,name):
        self.name = name
        self.users = {}

    def deposit(self, user, money):
        if user.name in self.users:
            # 기존의 돈
            self.users.update({user.name : self.users[user.name] + money})
            user.money -= money
        else:
            self.users.update({user.name : money})

    def show(self):
        print(self.users)


```

```python
class User:
    def __init__(self, name, money):
        self.name = name
        self.money = money

    def show(self):
        format_money = format(self.money, "_")
        print(f"{self.name}님의 잔액 : {format_money}원")
```

```python
# 클래스 메서드
class Korean:
    # 클래스 변수
    country = "한국"

    # 인스턴스X , 클래스변수 O
    @classmethod # @ : 데코레이터
    def trip(cls, country):
        if cls.country == country:
            print("국내여행")
        else:
            print("해외여행")

    # 정적 메서드
    @staticmethod
    def slogan():
        print("Imagine your korea")
```

```python
hong = Korean()

hong.trip("한국")
hong.slogan()

Korean.trip("미국")
Korean.slogan()
```

# 상속 (inheritance)

```python
class Person():
    def __init__(self,name):
        self.name = name

class Student(Person):
    def __init__(self, name, age):
        super().__init__(name)
        self.age = age
```

```python
hong = Person("홍길동")
jang = Student("장보고",20)

# print(hong)
# print(id(hong)) # 이놈이 해쉬 주소값


# 부모 -> 타입도 같이 전달!
# 모든 자식은 부모 타입이다.
# 부모는 자식타입이 아니다.

# 홍길동은 person타입이니?
print(isinstance(hong, Person))

# 홍길동은 student타입이니?
print(isinstance(hong, Student))


# 장보고는 Person 타입, Student 타입이니?
print(isinstance(jang, Person))
print(isinstance(jang, Student))
```

```python
class Animal:
    def __init__(self, name, feed, species):
        self.name = name
        self.feed = feed
        self.species = species

    def feeding(self):
        print(f"{self.name}은 먹이를 먹습니다.")

        

class Dog(Animal):
    def __init__(self, name, feed, species):
        super().__init__(name, feed, species)

    def act(self):
        print(f"{self.name}은 산책을 합니다")


        

class Cat(Animal):
    def __init__(self, name, feed, species):
        super().__init__(name, feed, species)

    def sleep(self):
        print(f"{self.name}은 잠을 자주 잡니다")
    
```

```python
dog = Dog("누렁이","잡식","강아지")
dog.act()
dog.feeding()

cat = Cat("야옹이","츄르","고양이")
cat.sleep()
cat.feeding()
```

```python
# 사람(부모), 학생(자식)
# 사람 - 이름, 나이, 키, 몸무게
# 학생 - 학생, 핸드폰

# 사람은 인사하는 메서드
# 학생은 공부하는 메서드

class Parent:
    def __init__(self, name, age, height, weight):
        self.name = name
        self.age = age
        self.height = height
        self.weight = weight

    @staticmethod
    def hello():
        print("안녕😁")

class Student(Parent):
    def __init__(self, name, age, height, weight, phone):
        super().__init__(name, age, height, weight)
        self.phone = phone

    
    def study(self):
        print(f"{self.name}은 공부를 합니다")
        
```

```python
parent = Parent("홍길동", 40 , "170cm", "50kg")
parent.hello()

student = Student("이순신", 10 , "180cm", "70kg", "01012341234")
student.study()
student.hello()
```

# 다형성

```python
class Car:
    def __init__(self, brand, price):
        self.brand = brand
        self.price = price

    def start_engine(self):
        print(f"{self.brand}는 열쇠로 시동을 켭니다")


    def stop_engine(self):
        print(f"{self.brand}는 열쇠로 시동을 끔니다")

class SuperCar(Car):

    def start_engine(self):
        print(f"{self.brand}는 음성으로 시동을 켭니다")


    def stop_engine(self):
        print(f"{self.brand}는 음성으로 시동을 끔니다")

    
```

```python
matiz = Car("마티즈", 1000000)
matiz.start_engine()
matiz.stop_engine()
```

```python
lamborghini = SuperCar("람보르기니", 50000000000)
lamborghini.start_engine()
lamborghini.stop_engine()
```

```python
# 휴대폰 필드
# 브랜드(brand), 가격(price)
# 2G 전화하기
# 3G 전화하기, 인터넷
# 4G 전화하기, 인터넷, 영상통화

class Phone2G:
    def __init__(self, brand, price):
        self.brand = brand
        self.price = price
        
    def network(self):
        print(f"{self.brand} 2G 입니다")

    def call(self):
        print(f"{self.brand} 전화 기능이 있습니다")

class Phone3G(Phone2G):
    def __init__(self, brand, price):
        super().__init__(brand, price)
        
    def network(self):
        print(f"{self.brand} 3G 입니다")

    def call(self):
        print(f"{self.brand} 전화 기능이 있습니다")

    def internet(self):
        print(f"{self.brand} 인터넷 기능이 있습니다")

class Phone4G(Phone3G):
    def __init__(self, brand, price):
        super().__init__(brand, price)
        
    def network(self):
        print(f"{self.brand} 4G 입니다")

    def call(self):
        print(f"{self.brand} 전화 기능이 있습니다")

    def internet(self):
        print(f"{self.brand} 인터넷 기능이 있습니다")    
        
    def videoCall(self):
        print(f"{self.brand} 영상통화 기능이 있습니다")    

        
```

```python
iphone = Phone2G("아이폰",100000)
galaxyS3 = Phone3G("갤럭시S3", 500000)
iphone12Pro = Phone4G("아이폰12프로", 10000000)
```

```python
iphone.network()
galaxyS3.network()
iphone12Pro.network()
```

```python
iphone12Pro.videoCall()
iphone12Pro.internet()
iphone12Pro.call()

galaxyS3.internet()
galaxyS3.call()

iphone.call()
```
